```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

def blur(img, mask, blur):
  fondo = cv2.GaussianBlur(img, (blur, blur), 0)
  fin = np.where(mask[:,:,None]>0, img, fondo)
  return fin

img = cv2.imread('./Img/rosa.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_,mask = cv2.threshold(gray, 100, 125, cv2.THRESH_BINARY_INV)
mask = cv2.bitwise_not(mask)
mask = cv2.GaussianBlur(mask, (5, 5), 0)

blurred = blur(img, mask, 15)
plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')
plt.subplot(122)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.title('Blurred')
plt.axis('off')
plt.show()

```

```python
img = cv2.imread('./Img/elementos.jpg')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_,binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

border,_ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cont = len(border)

plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(binary, cmap='gray')
plt.title(f'Conteo: {cont} elementos')
plt.axis('off')
plt.show()
```
