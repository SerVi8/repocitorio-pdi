```python
import cv2
import numpy as np

import matplotlib.pyplot as plt

# Cargar la imagen en escala de grises
image = cv2.imread('../Img/salamandra.png', cv2.IMREAD_GRAYSCALE)

# Gradiente Laplaciano
laplacian = cv2.Laplacian(image, cv2.CV_64F)

# Laplaciano Gaussiano (LoG)
# Primero aplicamos un filtro Gaussiano para suavizar la imagen
gaussian_blur = cv2.GaussianBlur(image, (5, 5), 0)
# Luego aplicamos el operador Laplaciano
laplacian_of_gaussian = cv2.Laplacian(gaussian_blur, cv2.CV_64F)

# Mostrar los resultados
plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1)
plt.title("Imagen Original")
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Gradiente Laplaciano")
plt.imshow(np.abs(laplacian), cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Laplaciano Gaussiano (LoG)")
plt.imshow(np.abs(laplacian_of_gaussian), cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
