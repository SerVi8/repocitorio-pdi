EJERCICIO 2

```python


def conuntRojos (img):
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  mask = cv2.inRange(img, (0, 0, 100), (100, 100, 255))
  img_gray = cv2.bitwise_and(img_gray, img_gray, mask=mask)
  _,binarizada = cv2.threshold(img_gray, 80,255,cv2.THRESH_BINARY_INV)
  contorno = measure.label(binarizada, connectivity=2)
  cont = np.max(contorno)
  return cont, binarizada
def conuntNaranjas (img):
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  mask = cv2.inRange(img, (0, 100, 180), (80, 170, 255))
  img_gray = cv2.bitwise_and(img_gray, img_gray, mask=mask)
  # plt.imshow(img_gray, cmap='gray')
  _,binarizada = cv2.threshold(img_gray, 130,255,cv2.THRESH_BINARY_INV)
  contorno = measure.label(binarizada, connectivity=2)
  cont = np.max(contorno)
  return cont, binarizada
def conuntVerde (img):
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  mask = cv2.inRange(img, (10, 150, 100), (100, 255, 180))
  img_gray = cv2.bitwise_and(img_gray, img_gray, mask=mask)
  # plt.imshow(img_gray, cmap='gray')
  _,binarizada = cv2.threshold(img_gray, 140,255,cv2.THRESH_BINARY_INV)
  contorno = measure.label(binarizada, connectivity=2)
  cont = np.max(contorno)
  return cont, binarizada
def conuntAmarillo (img):
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  mask = cv2.inRange(img, (30, 200, 220), (120, 240, 255))
  img_gray = cv2.bitwise_and(img_gray, img_gray, mask=mask)
  # plt.imshow(img_gray, cmap='gray')
  _,binarizada = cv2.threshold(img_gray, 195,255,cv2.THRESH_BINARY_INV)
  contorno = measure.label(binarizada, connectivity=2)
  cont = np.max(contorno)
  return cont, binarizada
def conuntAzul (img):
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  mask = cv2.inRange(img, (200, 100, 0), (255, 230, 140))
  img_gray = cv2.bitwise_and(img_gray, img_gray, mask=mask)
  # plt.imshow(img_gray, cmap='gray')
  _,binarizada = cv2.threshold(img_gray, 100,255,cv2.THRESH_BINARY_INV)
  contorno = measure.label(binarizada, connectivity=2)
  cont = np.max(contorno)
  return cont, binarizada

img = cv2.imread('colores.jpg')
num_red_objects1, binarizada1 = conuntRojos(img)
num_red_objects2, binarizada2 = conuntNaranjas(img)
num_red_objects3, binarizada3 = conuntAmarillo(img)
num_red_objects4, binarizada4 = conuntVerde(img)
num_red_objects5, binarizada5 = conuntAzul(img)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 7))
plt.subplot(1, 5, 1)
plt.imshow(binarizada1, cmap='gray')
plt.title(f'Objetos rojos detectados: {num_red_objects1}')
plt.subplot(1, 5, 2)
plt.imshow(binarizada2, cmap='gray')
plt.title(f'Objetos rojos detectados: {num_red_objects2}')
plt.subplot(1, 5, 3)
plt.imshow(binarizada3, cmap='gray')
plt.title(f'Objetos rojos detectados: {num_red_objects3}')
plt.subplot(1, 5, 4)
plt.imshow(binarizada4, cmap='gray')
plt.title(f'Objetos rojos detectados: {num_red_objects4}')
plt.subplot(1, 5, 5)
plt.imshow(binarizada5, cmap='gray')
plt.title(f'Objetos rojos detectados: {num_red_objects5}')
```

El filtro canny es un algoritmo popular que utiliza técnicas de suavizado, detección de gradientes y supresión de no-máximos para identificar bordes en una imagen.
Podemos usarlo para resaltar los bordes de los rostros, para luego aplicar técnicas de reconocimiento facial.

```python
img = cv2.imread('rostro.jpeg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
canny = cv2.Canny(img_gray,100,200)
plt.figure(figsize=(10,7))
plt.imshow (canny, cmap='gray')
```
