```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
imagen = cv2.imread('../Img/salamandra.png',cv2.IMREAD_GRAYSCALE)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
img_rest =cv2.erode(imagen,kernel)
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imagen,cmap='gray')

plt.subplot(122)
plt.imshow(img_rest,cmap='gray')
```

```python
imagen = cv2.imread('../Img/salamandra.png',cv2.IMREAD_GRAYSCALE)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
img_rest =cv2.dilate(imagen,kernel)
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imagen,cmap='gray')

plt.subplot(122)
plt.imshow(img_rest,cmap='gray')
```

```python
#EROCION BINARIA
imagen = cv2.imread('../Img/salamandra.png',cv2.IMREAD_GRAYSCALE)
_,img_bin=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY_INV)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
img_rest =cv2.erode(img_bin,kernel)#dilate
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imagen,cmap='gray')

plt.subplot(122)
plt.imshow(img_rest,cmap='gray')
```

```python
#APERTURA (EROSION + DILATACION)
imagen = cv2.imread('../Img/salamandra.png',cv2.IMREAD_GRAYSCALE)
_,img_bin=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY_INV)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(9,9))
apertura=cv2.morphologyEx(imagen,cv2.MORPH_OPEN,kernel)
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imagen,cmap='gray')

plt.subplot(122)
plt.imshow(apertura,cmap='gray')
```

```python
#CERRADO (DILATACION + ERCION)
imagen = cv2.imread('../Img/salamandra.png',cv2.IMREAD_GRAYSCALE)
_,img_bin=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY_INV)
#ELEMNETO ESTRUCTURANTE
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(9,9))
#aplicar erosion
#img_rest =cv2.erode(img_bin,kernel)#dilate
apertura=cv2.morphologyEx(imagen,cv2.MORPH_CLOSE,kernel)
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imagen,cmap='gray')

plt.subplot(122)
plt.imshow(apertura,cmap='gray')
## SKIMAHE => MORPHOLOGY.BINARY_OPENING
```

```python
from skimage import data , morphology
import matplotlib.pyplot as plt

img =  data.binary_blobs(length=64,n_dim=2)
kernel = morphology.disk(5)
apertura=morphology.binary_opening(img,kernel)
cerradura=morphology.binary_closing(img,kernel)

# plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(apertura,cmap='gray')
plt.subplot(122)
plt.imshow(cerradura,cmap='gray')
```
