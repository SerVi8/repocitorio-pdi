ALIASING
Este efecto se produce cuando cuando la imagen pierde calidad, se presenta cuand la cantidad de pixiles no pueden codificar todos los detalles de la imagen, el ojo umano lo percibe como una interferencia, por lo que se genera una distorcion en la imagen.

PATRONES DE MOIRE
Es un patron de interferencia que se muestra al superponer dos regillas de lineas delgadas con un cierto angulo. en las fotografias se produce que cuando hay patrones repetitivos de lineas o puntos.

```python
import cv2
import numpy as np
import matplotlib.pylab as plt

img1 = cv2.imread('./img/img1.jpg', 0)
img2 = cv2.imread('./img/img2.jpg', 0)
img3 = cv2.imread('./img/img3.jpg', 0)

maskfilterGauss = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,7,2],[2,7,11,7,1],[1,2,3,2,1]]) #kernel filtro de tipo de Gauss

img1_g = cv2.filter2D(img1,-1,maskfilterGauss/71)
img2_g = cv2.filter2D(img2,-1,maskfilterGauss/61)
img3_g = cv2.filter2D(img3,-1,maskfilterGauss/55)


plt.figure(figsize=(7,10))

plt.subplot(3,2,1)
plt.imshow(img1, cmap='gray')
plt.axis('off')
plt.subplot(3,2,2)
plt.imshow(img1_g, cmap='gray')
plt.axis('off')
plt.subplot(3,2,3)
plt.imshow(img2, cmap='gray')
plt.axis('off')
plt.subplot(3,2,4)
plt.imshow(img2_g, cmap='gray')
plt.axis('off')
plt.subplot(3,2,5)
plt.imshow(img3, cmap='gray')
plt.axis('off')
plt.subplot(3,2,6)
plt.imshow(img3_g, cmap='gray')
plt.axis('off')


```
