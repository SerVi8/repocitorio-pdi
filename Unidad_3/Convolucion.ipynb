```python
import cv2
import numpy as np
import matplotlib.pylab as plt
```

```python
img = cv2.imread('../Img/salamandra.png')
mask = np.array([[0,-1,0],[1,5,-1],[0,1,0]]) #kernel
img_conv = cv2.filter2D(img,-1,mask)
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img_conv, cmap='gray')
```

```python
img = cv2.imread('../Img/salamandra.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

kernel5x5 = np.ones((5,5),np.float32)/25 #suavizado
suavizado = cv2.filter2D(img_rgb,-1,kernel5x5) #imagen suavizada

kernel7x7 = np.ones((7,7),np.float32)/49 #mas suavizado
suavizado2 = cv2.filter2D(img_rgb,-1,kernel7x7) #imagen suavizada

kernel25x25 = np.ones((25,25),np.float32)/(25*25) #mas suavizado
suavizado3 = cv2.filter2D(img_rgb,-1,kernel25x25) #imagen suavizada

plt.figure(figsize=(20,20))

plt.subplot(1,4,1)
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')
plt.subplot(1,4,2)
plt.imshow(suavizado, cmap='gray')
plt.title('Suavizado 5x5')
plt.axis('off')
plt.subplot(1,4,3)
plt.imshow(suavizado2, cmap='gray')
plt.title('Suavizado 7x7')
plt.axis('off')
plt.subplot(1,4,4)
plt.imshow(suavizado3, cmap='gray')
plt.title('Suavizado 25x25')
plt.axis('off')
plt.show()
```

```python
img = cv2.cvtColor(cv2.imread('../Img/salamandra.png'), cv2.COLOR_RGB2BGR)
mask3 = np.array([[0,0,0],[-1,1,0],[0,0,0]])*21 #kernel realce de bordes
img_con3 = cv2.filter2D(img,-1,mask3)

mask4 = np.array([[-2,-1,0],[-1,1,1],[0,1,2]]) #kernel repujado
img_con4 = cv2.filter2D(img,-1,mask4)

mask5 = np.array([[0,1,0],[1,-4,1],[0,1,0]])*45 #kernel deteccion de bordes
img_con5 = cv2.filter2D(img,-1,mask5)

mask6 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])*21 #kernel filtro de tipo sobel
img_con6 = cv2.filter2D(img,-1,mask6)

mask7 = np.array([[1,-2,1],[-2,5,-2],[1,-2,1]]) #kernel filtro de tipo Shapen
img_con7 = cv2.filter2D(img,-1,mask7)

mask8 = np.array([[1,1,1],[1,-2,1],[-1,-1,-1]])*9 #kernel filtro norte
img_con8 = cv2.filter2D(img,-1,mask8)

mask9 = np.array([[-1,1,1],[-1,-2,1],[-1,1,1]])*9 #kernel filtro este
img_con9 = cv2.filter2D(img,-1,mask9)

mask10 = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,7,2],[2,7,11,7,1],[1,2,3,2,1]])/31 #kernel filtro de tipo de Gauss
img_con10 = cv2.filter2D(img,-1,mask10)

plt.figure(figsize=(20,12))

plt.subplot(3,3,1)
plt.imshow(img)
plt.title('Imagen Original')
plt.subplot(3,3,2)
plt.imshow(img_con3)
plt.title('Imagen Ralce de Bordes')
plt.subplot(3,3,3)
plt.imshow(img_con4)
plt.title('Imagen Repujado')
plt.subplot(3,3,4)
plt.imshow(img_con5)
plt.title('Imagen Detecion de Bordes')
plt.subplot(3,3,5)
plt.imshow(img_con6)
plt.title('Imagen Filtro de Tipo Sobel')
plt.subplot(3,3,6)
plt.imshow(img_con7)
plt.title('Imagen Filtro de Tipo Shapen')
plt.subplot(3,3,7)
plt.imshow(img_con8)
plt.title('Imagen Filtro Norte')
plt.subplot(3,3,8)
plt.imshow(img_con9)
plt.title('Imagen Filtro Este')
plt.subplot(3,3,9)
plt.imshow(img_con10)
plt.title('Imagen Filtro de tipo Gauss')
plt.show()
```

```python
import cv2 
import matplotlib.pyplot as plt
import numpy as np

def convolucion_media_horizontal(imagen,tamanioKernal):
    kernel=np.ones((1,tamanioKernal),np.float32)/tamanioKernal
    imgFin=cv2.filter2D(imagen,-1,kernel)
    return imgFin

path="./img//img4.jpg"

imagen=cv2.imread(path)
gris=cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
imagen_rest=convolucion_media_horizontal(gris,9)

plt.figure(figsize=(10,5))

plt.subplot(131)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))

plt.subplot(132)
plt.imshow(gris,cmap='gray')

plt.subplot(133)
plt.imshow(imagen_rest,cmap='gray')
plt.show()
```

```python
def convolucion_media_horizontal(imagen,tamanioKernal):
    kernel=np.ones((1,tamanioKernal),np.float32)/tamanioKernal
    imgFin=cv2.filter2D(imagen,-1,kernel)
    return imgFin

#path="../imagenes/dist.jpg"
path="../imagenes/Distorcion.png"

imagen=cv2.imread(path)
rgb=cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB)
imgRest_r=convolucion_media_horizontal(rgb[:,:,0],99)
imgRest_g=convolucion_media_horizontal(rgb[:,:,1],99)
imgRest_b=convolucion_media_horizontal(rgb[:,:,2],99)

imgColor=np.stack([imgRest_r,imgRest_g,imgRest_b],axis=2)


plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(rgb)

plt.subplot(122)
plt.imshow(imgColor)
plt.show()
```
